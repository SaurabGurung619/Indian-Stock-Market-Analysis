```python
import pandas as pd
import yfinance as yf

all_stock_symbol = [
    'TCS.NS', 'INFY.NS', 'RELIANCE.NS', 'ADANIENT.NS', 'HCLTECH.NS',
       'ITC.NS', 'BHARTIARTL.NS', 'BAJFINANCE.NS', 'LT.NS', 'TITAN.NS',
       'SUNPHARMA.NS', 'KOTAKBANK.NS', 'SBILIFE.NS', 'NTPC.NS', 'ONGC.NS',
       'HINDALCO.NS', 'ASIANPAINT.NS', 'DMART.NS'
]

empty_df = pd.DataFrame()

for i in all_stock_symbol:

    

    stock = yf.Ticker(i)

    data = stock.history(
        period="6y",
        interval="1d",
        auto_adjust=False
    )

    # Check if data exists
    if data.empty:
        print(f"No data found for {i}")
        continue

    # Convert index to Date column
    data = data.reset_index()

    # Add stock symbol
    data["Company_name"] = i

    # Add Dividends if missing
    if "Dividends" not in data.columns:
        data["Dividends"] = 0.0

    # Add Stock Splits if missing
    if "Stock Splits" not in data.columns:
        data["Stock Splits"] = 0.0

    # Keep only required columns
    data = data[
        [
            "Date",
            "Open",
            "High",
            "Low",
            "Close",
            "Volume",
            "Dividends",
            "Stock Splits",
            "Company_name"
        ]
    ]

    # Combine all stocks
    empty_df = pd.concat(
        [empty_df, data],
        ignore_index=True
    )

# Final dataset
stock_table = empty_df.copy()

stock_table
```
